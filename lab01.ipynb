## Символьный анализ и визуализация функции

Исследуем функцию: $g(x)=x^3+2x^2-5x-6$ на отрезке $[-4;4]$.

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Символьная переменная и функция
x = sp.symbols('x')
g = x**3 + 2*x**2 - 5*x - 6

print('Функция:')
print('g(x) =', g)

# Факторизация
g_factor = sp.factor(g)
print('\nФакторизация:')
print('g(x) =', g_factor)

# Точные корни
roots = sp.solve(g, x)
print('\nТочные корни:')
print(roots)

# Производная
derivative = sp.diff(g, x)
print('\nПроизводная:')
print("g'(x) =", derivative)

# Неопределённый интеграл
integral = sp.integrate(g, x)
print('\nНеопределённый интеграл:')
print('∫g(x)dx =', integral)
```

## Этап 4. Численный эксперимент и график

```python
# Преобразуем символьную формулу в функцию NumPy
g_num = sp.lambdify(x, g, 'numpy')

# Сетка из 401 точки на [-4; 4]
x_values = np.linspace(-4, 4, 401)
y_values = g_num(x_values)

# Точные корни в числовом виде
roots_float = [float(r) for r in roots]

# График
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='g(x) = x³ + 2x² − 5x − 6', linewidth=2)
plt.scatter(roots_float, [0]*len(roots_float), color='red', s=70,
            zorder=5, label='Точные корни')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

for r in roots_float:
    plt.annotate(f'x = {r:g}', (r, 0), xytext=(0, 10),
                 textcoords='offset points', ha='center')

plt.grid(alpha=0.3)
plt.xlabel('x')
plt.ylabel('g(x)')
plt.title('График функции g(x) на [-4; 4]')
plt.legend()
plt.show()
```

## Задание 4.2. Знаки функции

Корни: $-3$, $-1$, $2$.

$g(x)<0$ на $[-4;-3)$ и $(-1;2)$.

$g(x)>0$ на $(-3;-1)$ и $(2;4]$.

```python
# Проверка знаков подстановкой по одной точке из каждого интервала
test_points = [-4, -2, 0, 3]

print('Проверка знаков:')
for point in test_points:
    value = g.subs(x, point)
    print(f'g({point}) = {value}')
```

## Задание 4.3. Сетка из 21 точки

```python
# Меняем сетку с 401 на 21 точку
x_values_21 = np.linspace(-4, 4, 21)
y_values_21 = g_num(x_values_21)

plt.figure(figsize=(10, 6))
plt.plot(x_values_21, y_values_21, marker='o', label='g(x), 21 точка')
plt.scatter(roots_float, [0]*len(roots_float), color='red', s=70,
            zorder=5, label='Точные корни')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(alpha=0.3)
plt.xlabel('x')
plt.ylabel('g(x)')
plt.title('График функции при сетке из 21 точки')
plt.legend()
plt.show()

print('При 401 точке график получается более плавным и точным.')
print('При 21 точке график становится более грубым, так как между точками сетки меньше значений.')
```
