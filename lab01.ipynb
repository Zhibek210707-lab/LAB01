```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
g = x**3 + 2*x**2 - 5*x - 6

print('Функция:')
print('g(x) =', g)

g_num = sp.lambdify(x, g, 'numpy')

# 4.1. График с 401 точкой
x_values = np.linspace(-4, 4, 401)
y_values = g_num(x_values)

roots = sp.solve(g, x)
print('\nТочные корни:')
print(roots)

roots_float = [float(r) for r in roots]

plt.figure(figsize=(8, 4.5))
plt.plot(x_values, y_values, label='g(x)')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.scatter(roots_float, [0, 0, 0], color='red', s=60, label='Точные корни', zorder=5)
plt.xlabel('x')
plt.ylabel('g(x)')
plt.title('График функции g(x)')
plt.grid(alpha=0.3)
plt.legend()
plt.show()

# 4.2. Знаки функции
print('\nИнтервалы знакопостоянства:')

intervals = [(-4, -3), (-3, -1), (-1, 2), (2, 4)]

for a, b in intervals:
    point = (a + b) / 2
    value = g.subs(x, point)

    if value > 0:
        sign = 'положительна'
    else:
        sign = 'отрицательна'

    print(f'({a}; {b}): g({point}) = {value} -> функция {sign}')

print('\nПроверка подстановкой:')

test_points = [-3.5, -2, 0, 3]

for point in test_points:
    print(f'g({point}) =', g.subs(x, point))

# 4.3. График с 21 точкой
x_values_21 = np.linspace(-4, 4, 21)
y_values_21 = g_num(x_values_21)

plt.figure(figsize=(8, 4.5))
plt.plot(x_values_21, y_values_21, marker='o', label='g(x), 21 точка')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.scatter(roots_float, [0, 0, 0], color='red', s=60, label='Точные корни', zorder=5)
plt.xlabel('x')
plt.ylabel('g(x)')
plt.title('График функции g(x) - 21 точка')
plt.grid(alpha=0.3)
plt.legend()
plt.show()

print('\nСравнение графиков:')
print('При 401 точке график получается плавным и подробным.')
print('При 21 точке график становится более грубым.')
print('Чем плотнее сетка, тем точнее визуальное представление функции.')
```
